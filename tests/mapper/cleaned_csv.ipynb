```python
import pandas as pd
import pdfplumber
```

```python
PDF_FILE_PATH = "ayurveda.pdf"
```

```python
XLS_FILE_PATH = "ayurveda.xls"
```

```python
def extract_pdf(pdf_path):
    data = []
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            tables = page.extract_tables()
            for table in tables:
                for row in table:
                    if any(row):  # skip empty rows
                        data.append(row)
    return data
```

```python
extracted_data = extract_pdf(PDF_FILE_PATH)
len(extracted_data)
```

```python
tableHead = ['Term ID',
  'English term',
  'Description',
  'Sanskrit term in IAST*',
  'Sanskrit term']
```

```python
while tableHead in extracted_data:
    extracted_data.remove(tableHead)
```

```python
len(extracted_data)
```

```python
ayurveda_df = pd.DataFrame(extracted_data, columns=["Term ID", "English term", "Description", "Sanskrit (IAST)", "Sanskrit"])
```

```python
ayurveda_df
```

```python
ayurveda_df.drop(columns=["Term ID"], inplace=True)
```

```python
ayurveda_df = ayurveda_df.applymap(lambda x: x.lower() if isinstance(x, str) else x)
```

## Reading NAMASTE

```python
ayurveda_namaste = pd.read_excel('ayurveda.xls')
```

```python
ayurveda_namaste.drop(columns=['Sr No.'])
```

```python
cols_to_clean = ["NAMC_term","NAMC_term_diacritical", "NAMC_term_DEVANAGARI"]

for col in cols_to_clean:
    ayurveda_namaste[col] = ayurveda_namaste[col].astype(str).str.replace(r"\(.*?\)", "", regex=True).str.strip()

```

```python
ayurveda_namaste = ayurveda_namaste.applymap(lambda x: x.lower() if isinstance(x, str) else x)
```

```python
ayurveda_namaste
```

```python
merged_df = pd.merge(
    ayurveda_namaste,
    ayurveda_df,
    left_on='NAMC_term_diacritical',
    right_on='Sanskrit (IAST)',
    how='left'
)
```

```python
merged_df4 = pd.merge(
    ayurveda_namaste,
    ayurveda_df,
    left_on='NAMC_term',
    right_on='Sanskrit (IAST)',
    how='left'
)
```

```python
filtered_df4 = merged_df4[merged_df4['Term ID'].notna()].reset_index(drop=True)
```
